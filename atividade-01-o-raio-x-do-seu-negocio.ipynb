### ✍️ Sua justificativa (é ela que vale os 1,5)

**Por quê — explique a ordem em que o Python avalia os operadores no chunk
acima:**

> Passo 1 (`**`): A exponênciação é realizada primeiro pois ela possui prioridade em relação às outras operações, porém, isso só acontece quando ela está sendo aplicada em uma conta como a da projecao_b, já que a expressão não possui parênteses. Em contrapartida, na projecao_a, o uso dos parênteses faz com que as operações inseridas neles sejam priorizadas, sendo realizadas antes da exponênciação.
>
> Passo 2 (`*` e `/`): A multiplicação e a divisão são duas operações que seguem uma mesma regra de prioridades e são realizadas da esquerda para a direita, em que ambas não são realizadas antes de uma exponênciação mas, em contrapartida, possuem prioridade em relação à soma e subtração. Isso foi visto tanto na projecao_a quanto na projecao_b, nas quais as operações de adição só puderam ser executadas após as divisões/multiplicações serem feitas.
>
> Passo 3 (`+`): Por fim, é realizada a adição.
>
> Conclusão: A projecao_a representa corretamente o faturamento do negócio em 3 anos pois ao realizar a conta descrita levando em consideração as ordens de prioridade das operações em Python, o seu faturamento tende a crescer (realizando o que está contido nos parênteses, depois a exponênciação e, por fim, a multiplicação). A projecao_b dará um valor aproximado de 12879, valor quase igual ao inicial do faturamento. Ou seja, não houve crescimento significativo para que a projeção do faturamento esteja correto. A falta do parênteses na conta da projecao_b fez com que a exponênciação fosse realizada antes da divisão, da multiplicação e da adição e, logo em seguida, a multiplicação do MEU_FATURAMENTO por 1 diminuiu ainda mais a possibilidade de crescimento.

````python
# ▶️ Agora sim — rode e compare com a sua previsão
projecao_a = MEU_FATURAMENTO * (1 + MEU_CRESCIMENTO / 100) ** 3
projecao_b = MEU_FATURAMENTO * 1 + MEU_CRESCIMENTO / 100 ** 3

print(f"Faturamento hoje: R$ {MEU_FATURAMENTO:.2f}")
print(f"A: R$ {projecao_a:.2f}")
print(f"B: R$ {projecao_b:.2f}")
print(f"Diferença entre as duas: R$ {projecao_a - projecao_b:.2f}")

# TODO: substitua a nota abaixo — acertei? o que me surpreendeu?
registrar("previsao-revelada","Acertei. O que me surpreendeu foi a diferença entre as duas projeções, que é maior do que eu esperava. Mesmo que já tendo feito a conta da projeção b para descobrir o valor aproximado, eu não tinha noção de que o valor da projeção a seria tão maior do que o da projeção b.")
````

## 🥈 Exercício 2 — CAC vs. LTV: o negócio está saudável?

Duas siglas que todo fundador de negócio digital precisa saber de cor:

- **CAC** (Custo de Aquisição de Cliente) — quanto custa trazer 1 cliente
  novo. Você já tem: `MEU_CAC`.
- **LTV** (Lifetime Value) — quanto esse cliente vale ao longo de toda a
  relação com o negócio.

A aproximação padrão de unit economics:

$$\text{LTV} = \frac{\text{ticket médio mensal}}{\text{churn mensal}}$$

onde o churn entra em **fração**, não percentual — `MINHA_CHURN` chega em
percentual, converter é com você. (A lógica: se 1/10 dos clientes cancela
por mês, a vida útil esperada é de 10 meses.)

A regra de bolso que todo investidor de startup usa:

| Situação | Parecer |
|---|---|
| LTV > 3×CAC | negócio saudável |
| LTV ≤ 3×CAC | aquisição de cliente cara demais para o que ela retorna |

````python
# TODO 1: o LTV do seu negócio, em R$ (cuidado: churn em FRAÇÃO)
ltv = MEU_TICKET_MEDIO / (MINHA_CHURN / 100)

# TODO 2: a razão LTV/CAC
razao_ltv_cac = ltv / MEU_CAC

# TODO 3: um bool, sem if — LTV supera 3x o CAC?
saudavel = ltv > 3 * MEU_CAC

print(f"{MEU_NEGOCIO}: ticket R$ {MEU_TICKET_MEDIO:.2f}, churn {MINHA_CHURN:.2f}%")
print(f"  LTV .............. R$ {ltv:.2f}")
print(f"  CAC .............. R$ {MEU_CAC:.2f}")
print(f"  razão LTV/CAC ..... {razao_ltv_cac:.2f}x")
print("  Unit economics saudável?", saudavel)
````

````python
conferir(
    "exercicio-2",
    ltv=ltv,
    razao_ltv_cac=razao_ltv_cac,
    saudavel=saudavel,
)

# TODO: substitua a nota abaixo por uma frase sua sobre este exercício
registrar("exercicio-2", "Aprendi o que é LTV, CAC e a relação entre eles para saber se o negócio é saudável")
````

## 🥈 Exercício 3 — O crescimento de fato: CAGR do faturamento

O kit te deu **dois pontos** da série do faturamento:
`MEU_FATURAMENTO_BASE` no ano `MEU_ANO_BASE`, e `MEU_FATURAMENTO` no ano
`MEU_ANO`. Qual foi o crescimento médio por ano?

A resposta **não** é "crescimento total ÷ número de anos" — o crescimento
se capitaliza. A taxa média geométrica (*CAGR*) é:

$$\text{CAGR} = \left(\frac{V_{final}}{V_{inicial}}\right)^{1/n} - 1$$

> ⚠️ Conte os **intervalos**, não os anos: de 2024 a 2027 são **3**
> períodos de crescimento, não 4.

> 💡 O seu crescimento pode ter sido **negativo** — negócio digital jovem
> também encolhe, e a fórmula lida com isso naturalmente.

No fim, compare com o seu plano: `MEU_CRESCIMENTO`% ao ano é mais otimista
do que o negócio de fato entregou até aqui?

````python
# TODO 1: quantos ANOS de crescimento separam os dois pontos da série
anos_da_serie = MEU_ANO - MEU_ANO_BASE

# TODO 2: o crescimento total do período, em %
crescimento_total_pct = ((MEU_FATURAMENTO / MEU_FATURAMENTO_BASE) - 1) * 100

# TODO 3: o CAGR, em % ao ano (dica: expoente fracionário com **)
cagr_pct = ((MEU_FATURAMENTO / MEU_FATURAMENTO_BASE) ** (1 / anos_da_serie) - 1) * 100

# TODO 4: a média aritmética "ingênua" que muita gente reporta por engano
media_ingenua_pct = crescimento_total_pct / anos_da_serie

# TODO 5: um bool, sem if — o seu plano (MEU_CRESCIMENTO) é mais otimista
# do que o CAGR que o negócio realmente entregou?
projecao_otimista = MEU_CRESCIMENTO > cagr_pct

print(f"{MEU_NEGOCIO}, de {MEU_ANO_BASE} a {MEU_ANO} ({anos_da_serie} anos):")
print(f"  faturamento ........... R$ {MEU_FATURAMENTO_BASE:.2f} -> R$ {MEU_FATURAMENTO:.2f}")
print(f"  crescimento total ..... {crescimento_total_pct:.2f}%")
print(f"  CAGR (correto) ........ {cagr_pct:.2f}% ao ano")
print(f"  média ingênua (errada). {media_ingenua_pct:.2f}% ao ano")
print(f"  seu plano: {MEU_CRESCIMENTO}% ao ano")
print("  Plano mais otimista que o histórico?", projecao_otimista)
````

````python
conferir(
    "exercicio-3",
    anos_da_serie=anos_da_serie,
    crescimento_total_pct=crescimento_total_pct,
    cagr_pct=cagr_pct,
    media_ingenua_pct=media_ingenua_pct,
    projecao_otimista=projecao_otimista,
)

# TODO: substitua a nota abaixo por uma frase sua sobre este exercício
registrar("exercicio-3", "Essa questão foi mais difícil de interpretar, principalmente a parte de calcular o crescimento total e o CAGR, mas consegui entender a diferença entre eles e como calcular cada um corretamente.")
````

Todo pedido do seu negócio recebe um código como este:


Quatro campos, sempre nas mesmas posições: **categoria** (3 letras) — **ano**

O seu está em `MEU_CODIGO_PEDIDO`. Extraia os campos com **indexação e
fatiamento** — nada de digitar os valores à mão, mesmo que você "veja" o

## 🥇 Exercício 5 — A meta de faturamento

O seu plano de negócio fixou uma meta de faturamento mensal para
**2029** — é a variável `MINHA_META`. Partindo do seu ano foco (`MEU_ANO`)
e do crescimento projetado (`MEU_CRESCIMENTO`), a meta é alcançável?

Mesma fórmula de juros compostos da Parte 1:

$$F_{2029} = F_{\text{ano foco}} \times \left(1 + \frac{g}{100}\right)^{h}$$

onde $h$ é o horizonte, em anos, entre o ano foco e 2029 — contado como no
Exercício 3.

| Situação | Parecer |
|---|---|
| projeção ≥ meta | ✅ meta alcançada |
| entre 90% da meta e a meta | ⚠️ meta próxima, exige aceleração |
| abaixo de 90% da meta | ❌ meta inalcançável no ritmo atual |

````python
# TODO 1: o horizonte, em anos, entre o seu ano foco e 2029
horizonte = 2029 - MEU_ANO

# TODO 2: a projeção (atenção aos parênteses — lembre da Parte 1!)
faturamento_projetado = MEU_FATURAMENTO * (1 + MEU_CRESCIMENTO / 100) ** horizonte

# TODO 3: quanto isso representa da meta, em %
percentual_da_meta = (faturamento_projetado / MINHA_META) * 100

# TODO 4: o parecer, com if/elif/else
if faturamento_projetado >= MINHA_META:
    parecer = "✅ meta alcançada"
elif faturamento_projetado >= MINHA_META * 0.9:
    parecer = "⚠️ meta próxima, exige aceleração"
else:
    parecer = "❌ meta inalcançável no ritmo atual"

print(f"De {MEU_ANO} a 2029 são {horizonte} anos de crescimento.")
print(f"Faturamento projetado: R$ {faturamento_projetado:.2f}")
print(f"Isso é {percentual_da_meta:.1f}% da meta de R$ {MINHA_META:.2f}")
print(f"Parecer para {MEU_NEGOCIO}: {parecer}")
````

> ⚠️ **Espere — isto usa `if`, e você ainda não viu `if`!** Esta é a única
> exceção da atividade: o parecer de três faixas fica mais natural com uma
> cadeia condicional do que com aritmética booleana pura. Copie o padrão
> acima (`if` / `elif` / `else`, cada ramo atribuindo um texto) — não
> precisa entender a fundo ainda, a Aula 6 formaliza isso. Se preferir,
> pode resolver só com comparações e deixar o parecer para depois.

````python
conferir(
    "exercicio-5",
    horizonte=horizonte,
    faturamento_projetado=faturamento_projetado,
    percentual_da_meta=percentual_da_meta,
    parecer=parecer,
)

# TODO: substitua a nota abaixo por uma frase sua sobre este exercício
registrar("exercicio-5", "A questão foi bem tranquila, consegui calcular o faturamento projetado e o percentual da meta sem dificuldades. ")
````

Você financiou o investimento inicial do negócio (equipamento, reforma,
